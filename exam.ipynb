```python
# 🔹 Parent Classes (No imports used)
class Learner:
    def learn(self, topic):
        print(f"📘 {self.username} is learning {topic} through hands-on practice.")

class Collaborator:
    def collaborate(self, partner_name):
        print(f"🤝 {self.username} is collaborating with {partner_name} on a new project.")

# 🔹 Main User Profile Class
class UserProfile(Learner, Collaborator):
    def __init__(self, username, email, age, country, bio=""):
        self.username = username
        self.email = email
        self.age = age
        self.country = country
        self.bio = bio
        self.hobbies = []
        self.skills = []
        self.connections = []

    def introduce(self):
        print(f"\n👋 Hi, I’m {self.username} from {self.country}.")
        print(f"📧 Email: {self.email}")
        print(f"📝 Bio: {self.bio}")
        print(f"🎯 Hobbies: {', '.join(self.hobbies) if self.hobbies else 'None'}")
        print(f"💼 Skills: {', '.join(self.skills) if self.skills else 'None'}")
        print(f"🔗 Connections: {', '.join(self.connections) if self.connections else 'None'}")

    def add_hobby(self, hobby):
        self.hobbies.append(hobby)
        print(f"✅ Added hobby: {hobby}")

    def add_skill(self, skill):
        self.skills.append(skill)
        print(f"✅ Added skill: {skill}")

    def update_bio(self, new_bio):
        self.bio = new_bio
        print("✅ Bio updated!")

    def __add__(self, other):
        if isinstance(other, UserProfile):
            self.connections.append(other.username)
            other.connections.append(self.username)
            print(f"🔗 {self.username} and {other.username} are now connected!")
            return f"{self.username} + {other.username} = 👥 Developer team formed!"
        return "❌ Can only connect with another user."

    # 🔧 Show how the user does things
    def solve_problem(self, problem):
        print(f"🧠 {self.username} is solving: '{problem}' using logical steps and persistence.")

    def share_idea(self, idea):
        print(f"💡 {self.username} shares an idea: \"{idea}\" with the community.")

    def reflect(self):
        print(f"🪞 {self.username} reflects: What did I learn today? How can I improve tomorrow?")

    def full_profile(self):
        print("\n🧾----- FULL USER PROFILE -----")
        self.introduce()
        print("-------------------------------\n")

# 🔹 Example Usage
user1 = UserProfile("opeyemi", "opeyemi@email.com", 25, "Nigeria")
user2 = UserProfile("daniel", "daniel@email.com", 26, "Kenya")

user1.add_hobby("Reading tech blogs")
user1.add_skill("Python")
user1.update_bio("Full-stack dev passionate about building tools for developers.")
user1.learn("APIs")
user1.solve_problem("Fixing a login bug")
user1.share_idea("What if we had a zero-code backend generator?")
user1.reflect()

# Connect with another user
print(user1 + user2)

# Collaborate
user1.collaborate(user2.username)

# View final profile
user1.full_profile()
user2.full_profile()
```
