```python
# i admire two persons Tosin = BusinessM for Business and Kent for Software Developement = SoftwareM

class BusinessM:
    def businessMentor(self):
        return "Tosin is my Business Mentor and he loves sharing high revenue business tip to my mentees"
    
class SoftwareM:
    def softwareDevMen(self):
        return "Kent is my Software dev mentor and he loves to express himself through creating AI training modules and opening up internship opps to his community members."
    

    # Create the Autobiographical Class and Inherit from both Parents

class Mentee(BusinessM, SoftwareM):
    def __init__(self, name, age, location, hobbies):
        self.name = name
        self.age = age
        self.location = location
        self.hobbies = hobbies

    def __add__(self, other): # dunder
        if isinstance(other, Mentee):
            return f"{self.name} and {other.name} are having a Business implementation converstation!"
        return "Can only talk with another Mentee!"

    def introduce(self):
                return f"Hi, I’m {self.name}, {self.age} years old from {self.location}."

    def mentoring_class(self, topic):
                return f"{self.name} is learning about {topic}."

    def show_hobbies(self):
                return f"My hobbies are: {', '.join(self.hobbies)}"

me = Mentee("Opeyemi", 30, "Ibadan", ["python and AI agent", "reading", "music"])
friend = Mentee("Daniel", 24, "Abuja", ["AI agents and Business", "traveling"])

print(me.introduce())
print(me.mentoring_class("Python"))
print(me.show_hobbies())
print(me.businessMentor())
print(me.softwareDevMen())
print(me + friend)




  
```
