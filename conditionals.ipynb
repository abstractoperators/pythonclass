```python
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0:
            print(f"{n} equals {x} * {n//x}")
            break
```

```python
for num in range(2, 10):
    if num % 2 == 0:
        print(f"Found an even number {num}")
        continue
    print(f"Found an odd number {num}")
```

```python
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0
            print(f"{n} equals {x} * {n//x}")
            break
    else:
        # no break happened
        print(f"{n} is a prime number")
```
