```python
class Country:
    def __init__(self):
        self.stack =[]

    def push(self,element):
        self.stack.append(element)

    def pop(self):
        if self.isEmpty():
          return "Country is empty"
        return self.stack.pop()

    def peek(self):
        if self.isEmpty():
          return "Country is empty"
        return self.stack[-1]

    def isEmpty(self):
        return(self.stack) == 0
    def size(self):
        return len(self.stack)


myCountry = Country()
myCountry.push('Canada')
myCountry.push('Alberta')
myCountry.push('Manitoba')

print("Country:", myCountry.stack)
print("Pop:", myCountry.pop())
print("Peek:", myCountry.peek())
print("Size:", myCountry.size())
```
