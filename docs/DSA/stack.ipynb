```python
# stack using oop

class Currencies:
  def __init__(self):
    self.cur = []

  def push(self, element):
    self.cur.append(element)

  def pop(self):
    if self.isEmpty():
      return "Stack is empty"
    return self.cur.pop()

  def peek(self):
    if self.isEmpty():
      return "Stack is empty"
    return self.cur[-1]

  def isEmpty(self):
    return len(self.cur) == 0

  def size(self):
    return len(self.cur)
  


dbCurrencies = Currencies()

dbCurrencies.push("ngn")
dbCurrencies.push("gbp")
dbCurrencies.push("dol")
dbCurrencies.push("ghc")
dbCurrencies.push("eur")

print("Currencies: ", dbCurrencies.cur)
print("Pop: ", dbCurrencies.pop())
print("Stack after Pop: ", dbCurrencies.cur)
print("Peek: ", dbCurrencies.peek())
print("isEmpty: ", dbCurrencies.isEmpty())
print("Size: ", dbCurrencies.size())
```
