```python
#find the lowest value in a list

my_array = [7, 12, 9, 4, 11, 8]
my_array.sort() 

minVal = f"Lowest value: {my_array[0]}"
minVal

```

```python
my_array = [7, 12, 9, 4, 11, 8]
minVal = my_array[0] 

for i in my_array:
  if i < minVal:
    minVal = i

print('Lowest value:', minVal)
```

```python
my_array = [7, 12, 9, 4, 11, 8]
minVal = min(my_array)
print('Lowest value:', minVal)
```
