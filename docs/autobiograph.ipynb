```python
class Disciplined:
    def work_ethic(self):
        return "I am very disciplined and I work diligently."

class Innovative:
    def think_outside_box(self):
        return "I always look for innovative ways to solves tough problems."

class MyAutobiography(Disciplined, Innovative):
    def __init__(self, name, age, location, hobbies):
        self.name = name
        self.age = age
        self.location = location
        self.hobbies = hobbies

    def __add__(self, other):
        if isinstance(other, MyAutobiography):
            shared_hobbies = set(self.hobbies) & set(other.hobbies)
            shared = ', '.join(shared_hobbies) if shared_hobbies else 'none'
            return f"{self.name} and {other.name} are now very close buddies and share hobbies like {shared}!"
        return "I can only interact with someone like me."

    def introduce(self):
        return f"Hi, I’m {self.name}, {self.age} years old from {self.location}."

    def show_hobbies(self):
        return f"My hobbies are: {', '.join(self.hobbies)}."

    def daily_routine(self):
        return "I start my day with learning, then coding, and end with a short walk."

me = MyAutobiography("Tomike", 25, "Ibadan", ["coding", "reading", "music"])
friend = MyAutobiography("Philip", 26, "Lagos", ["music", "football", "reading"])


print(me.introduce())
print(me.show_hobbies())
print(me.daily_routine())
print(me.work_ethic())
print(me.think_outside_box())
print(me + friend)
```
