```python
class Staff:   
    def __init__(self,name,age,status):
        self.name =name
        self.age = age
        self.status = status
    def introduction(self):
        print("Hello, my name is " + self.status)

r1 = Staff ("Alice", 30, "Single")
r2 = Staff ("Peter", 50, "Married")
r1.introduction()
r2.introduction()
```

```python
class Person :
    def __init__(self,name,personality,isSitting):
        self.name = name
        self.personality = personality
        self.isSitting = isSitting

    def introduction(self):
        sittingStatus = 'sitting' if self.isSitting else 'standing'
        print(f"I am {self.name}  I am {self.personality}  I am currently {sittingStatus} ")

class Staff:   
    def __init__(self,name,age,status):
        self.name =name
        self.age = age
        self.status = status
    def introduction(self):
        print("Hello, my name is " + self.name)

r1 = Staff ("Alice", 30, "Single")
r2 = Staff ("Peter", 50, "Married")
r1.introduction()
r2.introduction()

p1 = Person("Tomike","Soft", True)
p1.introduction()

p2 = Person("Leke","Agressive", False)
p2.introduction()

p1.Staff = r2
p1.Staff.introduction()

p2.Staff = r1
p2.Staff.introduction()

```
